## 5 · The degree branch - what was tested, and what failed

Stage 1 says _whether_ to augment. Stage 2 says _which_ signal to build the role graph from: degree, Ψ or centrality.
The centrality rule is §2. This section is the search for a **degree** rule - some property of the original graph,
measurable before any training, that says "here, degree is the right signal".

Four families of ideas were tried. All four failed, and each failed for a different reason.

**1 · Generic graph properties.** The same 15 measurements the stage-1 rule and the centrality rule were found with -
size, density, clustering, homophily, degree spread, and so on. Screened twice, first on 29 augmenting graphs and then
on 40. The best candidate left 4 exceptions and its leave-one-out accuracy never beat simply guessing the commonest
answer. **No rule.**

**2 · Degree-specific properties.** Measurements built for this question alone: how many _distinct_ degree values a
graph has, how many nodes share a degree, how stable the degree role graph is between builds. One cut was frozen and
tested - `distinct_degrees < 158.5 ⇒ degree`. It scored **1/4**, and failed _in the wrong direction_: both graphs above
the cut, where it predicted Ψ, were actually won by degree.

**3 · DSC - Degree Structural Consistency.** A degree role graph connects nodes of similar degree, which only makes
sense if same-degree nodes really do occupy similar positions. DSC measures exactly that, and several variants were
tried. It produced the strongest correlation in the whole branch (ρ −0.433) - **but the sign flipped** once very
low-degree nodes were excluded, so it was not one stable measurement. What survives is descriptive, not a rule:
same-degree nodes are never very alike on _any_ panel graph, so degree's premise is weak everywhere.

**4 · DRD - Degree Role Dispersion.** If degree role edges help, perhaps it is because they link nodes that sit far
apart in the original graph. DRD measures whether same-degree nodes are unusually distant. Result: **DRD ≈ 1.0 on every
graph tested** - same-degree pairs are spread through the graph exactly like random pairs. No signal at all.

_Two further mechanisms were tested and also failed: whether low-degree nodes gain most from augmentation (they do not -
the effect runs the other way), and whether Ψ becomes redundant with degree (the measure turned out to be a proxy for
properties already screened and rejected)._

**Why they failed - and it is not a shortage of data.** Degree wins alone on only about 9 of 48 graphs, and on most
graphs the degree and Ψ role graphs are nearly identical, so the two cannot be told apart at all: 9 of the 20
degree-corpus graphs remain ties even under the tighter paired instrument. Full detail for every attempt is in
`docs/paper_log.md`.


```python
import pandas as pd
from virgo import frozen_rules as fr

# The eight validation cells, hard-coded from results/degree_family_scored.csv so the table renders without a re-run.
# Grouped by WHEN the call was made - the four pre-registered graphs first, then the four scored after training - and
# within each group the degree wins come first. Grouping this way keeps the two evidence classes visually separate.
VALID = pd.DataFrame([
    ("genetic_fission_yeast", 0.0014, "degree",     0.00040, 0.04, "before training"),
    ("ppi_yeast",             0.0140, "degree",     0.00030, 0.05, "before training"),
    ("ppi_human",             0.0299, "degree",     0.01530, 2.37, "before training"),
    ("ppi_fly",               0.0023, "centrality", 0.01052, 1.33, "before training"),
    ("ppi_mouse",             0.0483, "degree",     0.01420, 3.16, "after training"),
    ("github",                0.2907, "degree",     0.00060, 0.15, "after training"),
    ("twitch_ru",             0.6378, "degree",     0.00350, 0.71, "after training"),
    ("twitch_fr",             0.8036, "psi",        0.00080, 0.22, "after training"),
], columns=["dataset", "nbr_label_entropy", "winner", "lead", "sems", "when_predicted"])

# The call and its correctness come from the FROZEN artifact, never from a literal, so this table cannot drift from
# virgo/frozen_rules.py. A cell above the cut gets None: the rule makes no claim there, so there is nothing to score.
VALID["rule_says"] = [fr.predict_degree_onesided({"nbr_label_entropy": v}) for v in VALID.nbr_label_entropy]
VALID["correct"] = [None if r == "no claim" else w == r for r, w in zip(VALID.rule_says, VALID.winner)]

scored, unscored = VALID[VALID.correct.notna()], VALID[VALID.correct.isna()]
pre = scored[scored.when_predicted == "before training"]
clear = scored[scored.sems >= 1]

# Presentation only below this line. The predictor keeps its own name so the column states what the rule reads; `lead`
# stays in VALID as the underlying record but is not shown, since `how clear` already expresses it in noise units.
SHOW = {"dataset": "graph", "nbr_label_entropy": "label entropy", "rule_says": "rule says",
        "winner": "actual winner", "sems": "how clear", "correct": "correct?", "when_predicted": "when predicted"}
MARK = {True: "✓", False: "✗"}
GROUP_START = int((VALID.when_predicted == "after training").argmax())      # first row of the second group
BIGGER = [{"selector": "table", "props": [("font-size", "13px")]},          # this table only; STYLE stays 11px
          {"selector": "th", "props": [("font-size", "13px"), ("padding", "6px"),
                                       ("border-bottom", "2px solid #444"), ("font-weight", "600")]},
          {"selector": "td", "props": [("font-size", "13px"), ("padding", "6px")]},
          {"selector": "caption", "props": [("caption-side", "top"), ("text-align", "left"),
                                            ("font-weight", "600"), ("font-size", "13px"), ("padding", "8px 6px")]},
          {"selector": f"tbody tr:nth-child({GROUP_START + 1}) td", "props": [("border-top", "1px solid #999")]}]

show = VALID[list(SHOW)].rename(columns=SHOW)
CAPTION = (f"{fr.FROZEN_DEGREE_ONESIDED.predictor} &lt; {fr.FROZEN_DEGREE_ONESIDED.point} &rArr; degree"
           f" &nbsp;-&nbsp; {len(VALID)} datasets, {len(scored)} scored, {len(unscored)} no claim")

display(show.style
        .format({"label entropy": "{:.4f}", "how clear": "{:.2f} σ",
                 "correct?": lambda v: MARK.get(v, "-")})
        .set_properties(subset=["graph"], **{"text-align": "left", "font-weight": "600"})
        .set_properties(subset=["correct?"], **{"font-weight": "700"})
        .set_table_styles(STYLE + BIGGER)
        .set_caption(CAPTION)
        .hide(axis="index"))

print(f"{len(VALID)} datasets listed  |  {len(scored)} scored, {len(unscored)} not scored "
      f"({', '.join(unscored.dataset)}: above the cut, so the rule makes no claim)")
print(f"correct {int(scored.correct.sum())}/{len(scored)}   -   "
      f"before training {int(pre.correct.sum())}/{len(pre)}, "
      f"after training {int(scored.correct.sum() - pre.correct.sum())}/{len(scored) - len(pre)}")
print(f"of the {len(clear)} decided by more than 1 sem: {int(clear.correct.sum())}/{len(clear)} correct "
      f"({', '.join(clear.dataset)})")
```
